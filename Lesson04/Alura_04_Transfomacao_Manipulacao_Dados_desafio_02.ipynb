```python
import pandas as pd
df = pd.read_json('/content/dados_locacao_imoveis.json')
df.head()
```

```python
df = pd.json_normalize(df['dados_locacao'])
df.head()
```

```python
columns = list(df.columns)
df = df.explode(columns[1:])
df
```

```python
df.reset_index(inplace = True, drop = True)
df.head()
```

```python
df['valor_aluguel'] = df['valor_aluguel'].apply(lambda x : x.replace('$ ', '').replace(',', '.').replace(' reais', '').strip())
df.head()
```

```python
df.info()
```

```python
df['valor_aluguel'] = df['valor_aluguel'].astype(float)
df.head()
```

```python
df.info()
```

```python
df['apartamento'] = df['apartamento'].str.replace('(blocoAP)', '')
df['apartamento'] = df['apartamento'].str.strip()
# dados['apartamento'] = dados['apartamento'].str.replace(r' \(blocoAP\)', '', regex=True)
df.head()
```

```python
df.info()
```

```python
df['datas_combinadas_pagamento'] = pd.to_datetime(df['datas_combinadas_pagamento'], format='%d/%m/%Y')
df['datas_de_pagamento'] = pd.to_datetime(df['datas_de_pagamento'], format='%d/%m/%Y')
df.info()
```

```python
(df['datas_de_pagamento'] > df['datas_combinadas_pagamento'])
df[df['datas_de_pagamento'] > df['datas_combinadas_pagamento']]
```

```python
df['atrasos'] = (df['datas_de_pagamento'] - df['datas_combinadas_pagamento']).dt.days
df.head()
```

```python
df['atrasos'].mean()
```

```python
media_atraso_apt = df.groupby(df['apartamento'])[['atrasos']].mean()
media_atraso_apt
```
