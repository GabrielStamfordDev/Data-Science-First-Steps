```python
# The idea of the project is to study on real estate pricing
import pandas as pd
df = pd.read_json('/content/dados_hospedagem.json')
df.head()
```

```python
df = pd.json_normalize(df['info_moveis'])
df
```

```python
df.info()
```

```python
df[df['experiencia_local'] != '--']
```

```python
# That way, the column 'experiencia_local' is useless
df = df.drop(columns = 'experiencia_local')
```

```python
df
```

```python
columns = list(df.columns)
columns
```

```python
df = df.explode(columns[2:])
df
```

```python
df.reset_index(inplace = True, drop = True)
df
```

```python
colunas_inteiras = ['max_hospedes', 'quantidade_banheiros', 'quantidade_quartos', 'quantidade_camas']
df[colunas_inteiras] = df[colunas_inteiras].astype(int)
# Equivalent: df.loc[:, colunas_inteiras] = df.loc[:, colunas_inteiras].astype(int)
df.info()
```

```python
df['avaliacao_geral'] = df['avaliacao_geral'].astype(float)
df.info()
```

```python
colunas_preco = ['taxa_deposito', 'taxa_limpeza', 'preco']
df[colunas_preco] = df[colunas_preco].replace(r'\$|,|\s', '', regex=True)
df.head()

# Another way to do this convertion would be:
# df['preco'] = df['preco'].apply(lambda x : x.replace('$', '').replace(',', '').strip())
# This iterates through each element of the 'preco' column and treats it as if it were a normal string
# Than, to the others 2 columns:
# df[['taxa_deposito', 'taxa_limpeza']] = df[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x : x.replace('$', '').replace(',', '').strip())
```

```python
df[colunas_preco] = df[colunas_preco].astype(float)
df
```

```python
df.info()
```

```python
df['descricao_local'] = df['descricao_local'].str.lower()
# Basically str allows treating each element of a cell as a string and, because of that, using string methods
```

```python
df.head()
```

```python
df['descricao_local'][3169]
```

```python
df['descricao_local'] = df['descricao_local'].str.replace(r'[^a-zA-Z0-9\-\']', ' ', regex=True)
# Without 'r', python would treat '\' as an escape character
```

```python
df['descricao_local'][3169]
```

```python
df['descricao_local'] = df['descricao_local'].str.split()
df
```

```python
df['comodidades'] = df['comodidades'].replace(r'[{}"]', '' ,regex = True)
```

```python
df
```

```python
df['comodidades'] = df['comodidades'].str.split(',')
df
```

```python
df['descricao_vizinhanca'] = df['descricao_vizinhanca'].str.lower()
df['descricao_vizinhanca'] = df['descricao_vizinhanca'].replace(r'[^a-zA-Z0-9\-\'\s]', '', regex = True)
df['descricao_vizinhanca'] = df['descricao_vizinhanca'].str.replace('(?<!\w)-(?!\w)', '', regex=True)
df['descricao_vizinhanca'] = df['descricao_vizinhanca'].str.split()
df.head()
```
