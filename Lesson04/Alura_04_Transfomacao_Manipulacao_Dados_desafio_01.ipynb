```python
import pandas as pd
df = pd.read_json('/content/dados_vendas_clientes.json')
df.head()
```

```python
df = pd.json_normalize(df['dados_vendas'])
df.head()
```

```python
columns = list(df.columns)
df = df.explode(columns[1:])
df
```

```python
df.reset_index(inplace = True, drop = True)
```

```python
df
```

```python
df['Valor da compra'] = df['Valor da compra'].apply(lambda x : x.replace('R$ ', '').replace(',', '.').strip())
df
```

```python
df.info()
```

```python
df['Valor da compra'] = df['Valor da compra'].astype(float)
df.info()
```

```python
df['Cliente'] = df['Cliente'].str.lower()
df['Cliente'] = df['Cliente'].replace(r'[^a-z\s]', '' , regex = True)
df.head()
```

```python
df['Data de venda'] = pd.to_datetime(df['Data de venda'])
df.info()
```

```python
df.head()
```

```python
df['Data de venda'].dt.strftime('%m/%Y')
```

```python
compras_por_mes = df.groupby(df['Data de venda'].dt.strftime('%m/%Y'))[['Valor da compra']].sum()
compras_por_mes
```

```python
compra_por_cliente = df.groupby(df['Cliente'])[['Valor da compra']].sum()
compra_por_cliente
```
