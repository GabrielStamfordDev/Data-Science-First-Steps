```python
# Questão 3.1
import pandas as pd

url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv'
alunos = pd.read_csv(url)
```

```python
alunos
```

```python
alunos.isnull().sum()
```

```python
alunos.fillna(0, inplace=True)
alunos
```

```python
# Questão 3.2
alunos_sairam = alunos.query("Nome in ['Alice', 'Carlos']").index
```

```python
alunos.drop(alunos_sairam, inplace=True)
alunos
```

```python
# Questão 3.3
alunos.query("Aprovado==True")
```

```python
selecao = alunos["Aprovado"]
# selecao = alunos["Aprovado"] == True
alunos_aprovados = alunos[selecao]
```

```python
# Questão 3.4 e Extra
alunos_aprovados.to_csv("alunos_aprovados.csv", sep=";")
```

```python
alunos_aprovados = alunos_aprovados.replace(7, 8)
alunos_aprovados.to_csv("alunos_aprovados.csv", sep=",", index=False)
```

```python
alunos.drop(columns=['Pontos Extras', 'Notas Finais'], inplace=True)
```

```python
# Questão 4.1
alunos["Pontos_Extras"] = alunos["Notas"] * 0.4
alunos.head()
```

```python
# Questão 4.2
alunos["Notas_Finais"] = (alunos["Notas"] + alunos["Pontos_Extras"]).clip(upper=10)
alunos.head()
```

```python
# Questão 4.3
alunos["Aprovado_Final"] = alunos['Notas_Finais'].apply(lambda x: True if x >= 6 else False)
alunos.head()
```

```python
selecao = (alunos['Aprovado'] == False) & (alunos['Aprovado_Final'] == True)
alunos[selecao]
```
