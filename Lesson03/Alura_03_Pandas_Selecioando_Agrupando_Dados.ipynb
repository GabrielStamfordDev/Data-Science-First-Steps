```python
populacao = pd.read_excel('POP2022_Municipios.xls', header = 1, skipfooter = 34)
populacao.head()
```

```python
populacao.info()
```

```python
# Vai dar erro. A ideia é corrigir ele em seguida
populacao['POPULAÇÃO'] = populacao['POPULAÇÃO'].astype(int)
```

```python
populacao[populacao["POPULAÇÃO"].str.contains('\(', na=False)] # Sem o (, acharia q estava especificando um Regex entre ()
```

```python
populacao = populacao.assign(populacao_sem_parenteses = populacao['POPULAÇÃO'].replace('\(\d{1,2}\)' ,'',regex = True),
                populacao_tratada = lambda x : x.loc[:, 'populalcao_sem_parenteses'].replace('\.', '', regex = True))
```

```python
populacao[populacao["POPULAÇÃO"].str.contains('\(', na=False)]
```

```python
populacao = populacao.astype({'populacao_tratada':'int64'})
```

```python
populacao.grouppby('UF')[['POPULAÇÃO']].sum().reset_index()
```

```python
emissao_estados = emissions_per_year[emissions_per_year['Year'] == 2021].groupby('Estado')[['Emission']].sum().reset_index()
emissao_estados
```

```python
dados_agrupados = pd.merge(emissao_estados, populacao, left_on = 'Estado', right_on = 'UF')
dados_agrupados
```

```python
dados_agrupados.plot(x = 'populacao_tratada', y = 'Emission', kind = 'scatter', figsize = (10,6));
```

```python
import plotly.express as px
px.scatter(dataframe= dados_agrupados, x = 'populacao_tratada', y = 'Emission', text = 'Estado', opacity = 0)
```
